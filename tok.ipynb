```python
import tiktoken
```

```python
tokenizer = tiktoken.encoding_for_model(model_name="gpt-4")
```

```python
text = "Who is Pm of India ?"
```

```python
tokenIds = tokenizer.encode(text)
```

```python
tokenIds
```

```python
dec = tokenizer.decode(tokens=tokenIds)
```
