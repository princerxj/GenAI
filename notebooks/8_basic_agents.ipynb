```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a : int, b : int) : 
    """
        It will return the sum of two numbers 

        Args : 
         a : Number One 
         b : Number two 
         
    """
    return a + b 
```

```python
@tool
def multiply_numbers(a : int, b : int) : 
    """
        It will return the product of two numbers 

        Args : 
         a : Number One 
         b : Number two 
         
    """
    return a * b 
```

```python
add_numbers.invoke({"a" : 12, "b" : 3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model = "qwen/qwen3.8-27b")
```

```python
agent = create_agent(
    model = llm , 
    tools = [add_numbers, multiply_numbers],
    system_prompt="You are a maths teacher , and always use tool for calculation"
)
```

```python
response = agent.invoke({"messages" : [{"role" : "user", "content" : "What is 6 +  2 * 3 ?"}]})
```
