```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel) : 
    name : str 
    email : str 
    age : int 
```

```python
user = User(name="Prince", email="pr23@gmail.com", age="iod")
```

##### Field Validation

```python
class Product(BaseModel) : 
    title : str = Field(max_length=10)
    price : int = Field(gt=0, le = 100)
    desc : str = Field(min_length=30)
```

```python
prod1 = Product(title="Soap", price=10, desc="Best Soap in the iNdian market")
```

```python
prod1.title
```

```python
prod2 = Product(title="Ni Washing Powder", price = 0, desc="Bad Washing Powder")
```

```python
class Order(BaseModel) : 
    id : str = Field(min_length=6, max_length=12)
    product : Product
```

```python
o1 = Order(id="uifhfdhjk", product=Product(title="Soap", price=12, desc="Best soap all over the world, used by vereyone across the globe"))
```

```python
o1
```

```python
o1.product.title
```

```python
class Carts(BaseModel) : 
    id : str = Field(min_length=6, max_length=12)
    items : List[Product]
    comment : str = Field(default="No Comment")
```

```python
cart = Carts(id = "8793r2hjf", items = [o1.product, prod1])
```

```python
cart.items
```

### Annotated - Gives permission to make custom validator

```python
common_validator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel) : 
    id : str = Field(min_length=6, max_length=12)
    items : common_validator
    comment : str = Field(default="No Comment")
```

```python
class Order(BaseModel) : 
    products : common_validator
    id : str
```

```python
order = Order(id="uifhfdhjk", products=[Product(title="Soap", price=12, desc="Best soap all over the world, used by vereyone across the globe")])
```
